# Creation de video au traitement similaire pour test de généralisation


```python
import cv2
import random
import os
```

```python
# Fonction pour le shuffle des frames
def shuffle_frames(frames):
    random.shuffle(frames)
    return frames
```

```python
# Fonction pour la creation de nouvelle video
def create_new_video(frame_list, output_path, frame_width, frame_height, fps):
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))

    for frame in frame_list:
        out.write(frame)

    out.release()
```

```python
# lecture de la video original
original_video_path = 'original.mp4'
cap = cv2.VideoCapture(original_video_path)
```

```python
#Caracteristiques de la video
frame_width = int(cap.get(3))
frame_height = int(cap.get(4))
fps = int(cap.get(5))
```

```python
# Extraction de frames
frames = []
while True:
    ret, frame = cap.read()
    if not ret:
        break
    frames.append(frame)

cap.release()
```

```python
# Shuffle frames
shuffled_frames = shuffle_frames(frames)

# Insertion de frames intrus
unrelated_frames = [cv2.imread('corruption1.jpeg'), cv2.imread('corruption2.jpg')]

# Combinaison de frames
all_frames = shuffled_frames + unrelated_frames
all_frames=shuffle_frames(all_frames)
# Creation de la video corrompue
output_video_path = 'shuffled_video.mp4'
create_new_video(all_frames, output_video_path, frame_width, frame_height, fps)

#message de succes
print(f"New video created: {output_video_path}")
```
